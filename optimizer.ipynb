## Orbital Trajectory Optimizer
##### This program will take in user defined orbital elements, propagate the spacecraft orbits using hapsira and astropy, compute maneuver costs, and optimize using Scipy. The optimizer will find the minimimum-cost solution validated against STK or bsk models.

```python
### Import hapsira and astropy libraries. Will be pruned when the code is finalized.
from hapsira.bodies import Earth, Moon, Sun
from hapsira.twobody import Orbit
from hapsira.plotting import OrbitPlotter
from hapsira.maneuver import Maneuver
from hapsira.plotting.orbit.backends import Matplotlib2D, Plotly3D
from hapsira.ephem import Ephem
from hapsira.twobody.sampling import EpochBounds
from astropy import units as u
from astropy.time import Time
from astropy.coordinates import solar_system_ephemeris
```

## User Defined Orbital Elements
##### The classical Keplerian orbital elements for the initial and final orbits are defined below. These can be modified to whichever values the user would like. The units are applied using `astropy.units`.

```python
    # Radius of the Earth in kilometers
    r = 6371 << u.km

    # Define the initial orbit parameters: semi-major axis, eccentricity, inclination, right ascension of ascending node, argument of periapsis, and true anomaly.
    a_i = (800 << u.km) + r
    e_i = 0 << u.dimensionless_unscaled
    inc_i = 0 << u.deg
    raan_i = 0 << u.deg
    argp_i = 0 << u.deg
    nu_i = 0 << u.deg

    # Define the final orbit parameters: semi-major axis, eccentricity, inclination, right ascension of ascending node, argument of periapsis, and true anomaly.
    a_f = (10000 << u.km) + r
    e_f = 0 << u.dimensionless_unscaled
    inc_f = 0 << u.deg
    raan_f = 0 << u.deg
    argp_f = 0 << u.deg
    nu_f = 0 << u.deg
```

## Orbit Propagation
##### The orbits get propagated using the elements defined above and `Orbit_from_classical`. They're then plotted onto 3D plots using `OrbitPlotter()` and `Plotly3D`.

```python
orb_i = Orbit.from_classical(Earth, a_i, e_i, inc_i, raan_i, argp_i, nu_i)
orb_f = Orbit.from_classical(Earth, a_f, e_f, inc_f, raan_f, argp_f, nu_f)

op = OrbitPlotter(backend=Plotly3D())
op.plot(orb_i, label="Initial Orbit")
op.plot(orb_f, label="Final Orbit")
op.show()
```
